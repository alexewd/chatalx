```python
import io
from lars import iis, csv
from lars import apache

with io.open('data/access.log', 'r') as f, \
         io.open('data/output_log.csv', 'wb') as outfile:
    with apache.ApacheSource(f) as source, csv.CSVTarget(outfile) as target:
        for row in source:
            target.write(row)
```
